**Workspace Extraction & Module Verification**

Description: Unpacks the Session 5 repository ZIP file directly into the active directory and audits the workspace to ensure all updated modules are accounted for.

```python
import os
import zipfile

drive_path = "/content/drive/MyDrive/audit_llm"

if os.path.exists(drive_path):
    %cd {drive_path}

    # 1. Look for a ZIP file to extract
    zip_files = [f for f in os.listdir('.') if f.endswith('.zip')]
    if zip_files:
        target_zip = zip_files[0]
        print(f"📦 Found ZIP archive: {target_zip}. Extracting to workspace...")
        with zipfile.ZipFile(target_zip, 'r') as zip_ref:
            zip_ref.extractall('.')
        print("✅ Extraction complete.")
    else:
        print("ℹ️ No ZIP file found. Assuming individual .py files were uploaded directly.")

    # 2. List the active Python files in the workspace to confirm they are present
    py_files = [f for f in os.listdir('.') if f.endswith('.py')]
    print(f"\n📂 Active Python modules in workspace ({len(py_files)} found):")
    for file in sorted(py_files):
        print(f"  - {file}")

else:
    print(f"❌ Error: Drive folder not found at {drive_path}. Please ensure your Drive is mounted.")
```
